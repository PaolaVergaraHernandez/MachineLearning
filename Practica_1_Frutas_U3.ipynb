```python
# Importar librerias
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, confusion_matrix
frutas = pd.read_csv("frutas.csv")
frutas.head()
```

```python
X = frutas[["height", "width", "mass", "color_score"]]

y= frutas["fruit_name"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
clasificacion_KNN = KNeighborsClassifier(n_neighbors=1)
```

```python
clasificacion_KNN = KNeighborsClassifier(n_neighbors=1)
clasificacion_KNN.fit(X_train, y_train)
print("Precision del clasificador en fase de entrenamiento",
      clasificacion_KNN.score(X_train, y_train))
```

```python
y_pred = clasificacion_KNN.predict(X_test)
print("Reporte de métricas del clasificador: \n", classification_report(y_test, y_pred))

print ("Precisión:", clasificacion_KNN.score(X_test, y_test))
```

```python
print ("Matriz de confusión: \n", confusion_matrix(y_test, y_pred), "\n")
```

```python
from sklearn.inspection import permutation_importance
result = permutation_importance(clasificacion_KNN, X_test, y_test, n_repeats=30,
                                random_state=42)

importancia = pd.DataFrame({
    "Variable": X.columns,
    "Importancia": result.importances_mean,
    "Desviación": result.importances_std,
}).sort_values(by="Importancia", ascending=False)
print(importancia)
```

```python
import plotly.express as px
fig = px.bar(importancia, x="Variable", y="Importancia", error_y="Desviación", title="Importancia de características (KNN - Permutation Importance)",
text_auto=".3f", color= "Variable")
fig.update_layout(width=600,height=400)
fig.show()
```

```python
from sklearn.preprocessing import StandardScaler

X = frutas[["height", "width", "mass", "color_score"]]
y = frutas["fruit_name"]

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42, stratify=y)

# Inicializar y aplicar el escalador
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Entrenar el modelo con los datos ESCALADOS
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train_scaled, y_train)

# Realizar las predicciones con los datos de prueba escalados
y_pred_scaled = knn.predict(X_test_scaled)

print("\nReporte de clasificacion:")
print(classification_report(y_test, y_pred_scaled))
```

```python
print ("Matriz de confusion: \n", confusion_matrix(y_test, y_pred_scaled), "\n")
```

```python
from sklearn.inspection import permutation_importance

# Usamos el modelo 'knn' que fue el que entrenamos con datos escalados
result = permutation_importance(knn, X_test_scaled, y_test, n_repeats=30,
                                random_state=42)

importancia = pd.DataFrame({
    "Variable": X.columns,
    "Importancia": result.importances_mean,
    "Desviación": result.importances_std,
}).sort_values(by="Importancia", ascending=False)

importancia
```

```python
import plotly.express as px
fig4 = px.bar(importancia, x="Variable", y="Importancia",
              error_y="Desviación", title="Importancia de características (KNN - Permutation Importance)",
text_auto=".3f", color= "Variable")
fig4.update_layout(width=600,height=400)
fig4.show()
```
